```python
from sqlalchemy import create_engine
from kaggle_secrets import UserSecretsClient

# 1. Fetch your connection string securely from Kaggle Secrets
user_secrets = UserSecretsClient()
db_url = user_secrets.get_secret("DB_URL")

# Ensure standard SQLAlchemy driver prefix
if db_url.startswith("postgres://"):
    db_url = db_url.replace("postgres://", "postgresql://", 1)

# 2. Create the engine and push the DataFrame to Neon PostgreSQL
engine = create_engine(db_url)

print("Connecting to Neon PostgreSQL...")
```

```python
#splittin the main csv into four different tables
dim_customers=df1[['customer_id','customer_fname','customer_lname','customer_segment',
                   'customer_email','customer_password','customer_street','customer_zipcode','customer_city',
                   'customer_state','customer_country']].drop_duplicates(subset=['customer_id']).copy()
```

```python
dim_products=df1[['product_card_id','product_category_id','product_name',
                 'product_price',
                  'category_id','category_name','department_id','department_name']
].drop_duplicates(subset=['product_card_id']).copy()
```

```python
dim_logistics = df1[['shipping_mode',
                     'delivery_status', 
                     'late_delivery_risk',
                     'market', 'order_region']].drop_duplicates().reset_index(drop=True).copy()
#creating logistics id and giving it a index values
dim_logistics['logistics_id'] = dim_logistics.index + 1
```

```python
orders_cols=[c for c in['order_item_id', 'order_id', 'customer_id', 'product_card_id', 'logistics_id',
                        'order_item_quantity', 'order_item_product_price', 'order_item_discount',
                        'order_item_discount_rate', 'order_item_total', 'order_item_profit_ratio',
                        'sales', 'sales_per_customer', 'benefit_per_order', 'order_profit_per_order']
            if c in df1.columns]
dim_orders_items=df1[orders_cols].drop_duplicates(subset=['order_item_id']).copy()
```

```python
dim_orders=df1[['order_id', 'order_date', 'shipping_date', 'order_status', 
                  'type', 'market', 'order_region', 'order_country', 
                  'order_state', 'order_city', 'order_zipcode', 
                  'days_for_shipping', 'days_for_shipment']].drop_duplicates(subset=['order_id']).copy()
```

```python
#merging table dim logistic with df1 raw table for the following columns using left merge
df1=df1.merge(dim_logistics,on=['shipping_mode',
                                'delivery_status','late_delivery_risk'],how='left')
```

```python
#pushing the tables on the NEON db
print("pushing optimized tables on neon")
dim_customers.to_sql('dim_customers',con=engine,if_exists='replace',index=False)
dim_products.to_sql('dim_products',con=engine,if_exists='replace',index=False)
dim_logistics.to_sql('dim_logistics',con=engine,if_exists='replace',index=False)
print("pushed")
```

```python
orders_items_cols = [
    "order_id",
    "customer_id",
    "product_card_id",
    "logistics_id",
    "order_item_quantity",
    "order_item_product_price",
    "order_item_discount",
    "order_item_discount_rate",
    "order_item_total",
    "order_item_profit_ratio",
    "sales",
    "sales_per_customer",
    "benefit_per_order",
    "order_profit_per_order"
]

dim_orders_items = df1[orders_items_cols].copy()

# Keep the existing row structure and create a unique item key
dim_orders_items["item_seq"] = (
    dim_orders_items.groupby("order_id").cumcount() + 1
)

dim_orders_items["order_item_key"] = (
    dim_orders_items["order_id"].astype(str)
    + "_"
    + dim_orders_items["item_seq"].astype(str)
)

dim_orders_items = dim_orders_items.drop(columns=["item_seq"])
```

```python
print("Orders:", df1["order_id"].nunique())
print("Order item rows:", len(dim_orders_items))
print("Unique item keys:", dim_orders_items["order_item_key"].nunique())
```

```python
df1 = df1.rename(columns={
    "market_x": "customer_market",
    "market_y": "order_market",
    "order_region_x": "customer_region",
    "order_region_y": "order_region"
})
```

```python
dim_orders = df1[
    [
        "order_id",
        "order_date",
        "shipping_date",
        "order_status",
        "type",

        "customer_market",
        "order_market",
        "customer_region",
        "order_region",

        "order_country",
        "order_state",
        "order_city",
        "order_zipcode",

        "days_for_shipping",
        "days_for_shipment"
    ]
].drop_duplicates(subset=["order_id"]).copy()
```

```python
from sqlalchemy import create_engine, text
from kaggle_secrets import UserSecretsClient

user_secrets = UserSecretsClient()
db_url = user_secrets.get_secret("DB_URL")

if db_url.startswith("postgres://"):
    db_url = db_url.replace("postgres://", "postgresql://", 1)

engine = create_engine(
    db_url,
    pool_pre_ping=True,
    pool_recycle=300
)

print("New Neon engine created.")
with engine.connect() as conn:
    result = conn.execute(text("SELECT 1"))
    print(result.scalar())

print("Neon connection successful.")
```

```python
print("Uploading dim_orders...")

dim_orders.to_sql(
    "dim_orders",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=5000
)

print("dim_orders uploaded.")
print("Uploading dim_orders_items...")

dim_orders_items.to_sql(
    "dim_orders_items",
    con=engine,
    if_exists="replace",
    index=False,
    chunksize=5000
)

print("dim_orders_items uploaded.")
```

```python
tables = {
    "dim_customers": dim_customers,
    "dim_products": dim_products,
    "dim_logistics": dim_logistics,
    "dim_orders": dim_orders,
    "dim_orders_items": dim_orders_items
}
```

```python
for name, df in tables.items():

    print("\n" + "=" * 60)
    print(name)
    print("=" * 60)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))
    print("Duplicate rows:", df.duplicated().sum())
    print("Missing values:", df.isnull().sum().sum())

    print("\nColumns:")
    print(df.columns.tolist())
```
